The primary target would be a **relevance score** for a given `(user, product)` pair. This score is typically a proxy, derived from observed user behavior. ## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
# NDCG is widely used in information retrieval and recommendation systems.
# It penalizes relevant items that appear lower in the search results.
print("Success Metric: NDCG@K (Higher is better)")
```

The most appropriate success metric for a ranking task is **Normalized Discounted Cumulative Gain (NDCG) at K**. A higher NDCG@K indicates a better model, as it assigns higher scores if highly relevant items appear earlier in the ranked list. This metric accounts for both the relevance of items and their position in the list, which aligns perfectly with the goal of presenting the most relevant items first.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
# A fixed rule would quickly become unwieldy and ineffective in capturing the multifaceted nature of user-product relevance.
# Machine learning allows for automated pattern discovery and adaptation.
print("ML excels due to complexity, dynamism, and scalability over fixed rules.")
```

Machine learning beats a fixed rule here because product relevance is a highly complex and dynamic phenomenon driven by numerous interacting factors. A fixed rule (e.g., 'always recommend top-selling products' or 'recommend products from the user's last purchased category') would fail to capture:
1.  **Nuance and Subtlety**: User preferences are not static; they change with context (time of day, device), seasonality, and evolving trends.
2.  **Interaction Effects**: The impact of one feature (e.g., user's age) on relevance might depend on another feature (e.g., product category), which is hard to encode in simple `if-else` statements.
3.  **Scalability**: Manually crafting and maintaining rules for millions of users and products, and their combinations, is intractable.
4.  **Adaptability**: ML models can learn from new data, continuously improving their predictions and adapting to shifts in user behavior or product popularity, something fixed rules cannot do without constant manual updates.
