## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

Here's how I'd sort the data fields:

**Features (What we use to predict):**
*   `search_query`: The actual words someone typed.
*   `device_type`: Like 'mobile', 'desktop', or 'tablet'.
*   `country_code`: Which country the search came from (e.g., 'US', 'GB').
*   `query_length_chars`: How many letters are in the search query.
*   `num_words_in_query`: How many words are in the search query.
*   `time_of_day_hour`: The hour of the day the search happened.
*   `day_of_week`: Which day of the week the search happened.
*   `search_volume_last_7_days`: How popular this search was recently.
*   `query_category`: What kind of thing the search is about (e.g., 'electronics', 'food').

**Labels (What we want to predict or measure):**
*   `total_impressions`: How many times the search results were shown.
*   `total_clicks`: How many times someone clicked on a search result.
*   `click_through_rate`: The percentage of impressions that got clicks.
*   `total_conversions`: How many times a search led to a desired action (like a purchase).
*   `conversion_rate`: The percentage of clicks that led to a conversion.

**Context (Extra info, not directly a feature or label):**
*   `date`: The day the search happened.
*   `raw_log_timestamp`: The exact time the search happened.

**Excluded (Fields we won't use and why):**
*   `user_id_hash`: This is like a secret ID for a user. We don't need it because we are looking at overall trends, not individual people. It's also private.
*   `ip_address`: This tells us where a computer is located. It's also private and not needed for our general analysis.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This dataset has some limitations:

*   **Short Timeframe**: It only covers January 2023. This means we can't see how search behavior changes over seasons, holidays, or longer periods throughout the year.
*   **Search-Only Data**: The data only focuses on search queries. It doesn't tell us about other ways users might find information or interact with a product, like social media, direct website visits, or email marketing.
*   **No Personal Identifiers**: Because `user_id_hash` and `ip_address` are excluded for privacy, we can't track individual user journeys or understand personalized search trends.
*   **Specific Geographical Coverage**: The data might only cover certain regions or countries, so it may not represent global search behavior.
